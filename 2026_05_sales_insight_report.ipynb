```python
# ================================================
# SALES INSIGHT REPORT — MINI MARKET MAY 2026
# Analyst: [Your Name]
# Date: 2026-05-[today]
# ================================================

import pandas as pd
import matplotlib.pyplot as plt

data = {
    'date': ['2026-05-05', '2026-05-06', '2026-05-07',
             '2026-05-08', '2026-05-09'],
    'sales_amount': [15000, 8000, 22000, 11000, 18000],
    'customer_type': ['Walk-in', 'Regular', 'Walk-in',
                      'Regular', 'Walk-in'],
    'product_category': ['Food', 'Drinks', 'Food',
                         'Household', 'Food']
}

df = pd.DataFrame(data)
print("Data loaded! Rows:", len(df))
```

```python
# ================================================
# SECTION 1: COMPLETE STATISTICS SUMMARY
# ================================================

total = df["sales_amount"].sum()
average = df["sales_amount"].mean()
median = df["sales_amount"].median()
highest = df["sales_amount"].max()
lowest = df["sales_amount"].min()
sales_range = highest - lowest
std = df["sales_amount"].std()
top_20_threshold = df["sales_amount"].quantile(0.8)

print("=" * 45)
print("   SALES INSIGHT REPORT — MAY 2026")
print("=" * 45)
print(f"Total Revenue:      ₦{total:,.0f}")
print(f"Daily Average:      ₦{average:,.0f}")
print(f"Median (typical):   ₦{median:,.0f}")
print(f"Highest Day:        ₦{highest:,.0f}")
print(f"Lowest Day:         ₦{lowest:,.0f}")
print(f"Range:              ₦{sales_range:,.0f}")
print(f"Std Deviation:      ₦{std:,.0f}")
print(f"Top 20% Threshold:  ₦{top_20_threshold:,.0f}")
print("=" * 45)
```

```python
# ================================================
# SECTION 2: WHAT-IF SCENARIOS
# ================================================

print("WHAT-IF SCENARIOS:")
print("-" * 45)

# What-if 1: Every day performed at median level
median_week = median * 5
print(f"1. If ALL 5 days hit median (₦{median:,.0f}):")
print(f"   Weekly revenue = ₦{median_week:,.0f}")
print(f"   vs actual ₦{total:,.0f}")
print(f"   Difference: ₦{median_week - total:,.0f}")
print()

# What-if 2: Worst day improved by 20%
improved_worst = lowest * 1.20
improvement_gain = (improved_worst - lowest)
print(f"2. If worst day (₦{lowest:,.0f}) improved by 20%:")
print(f"   New worst day = ₦{improved_worst:,.0f}")
print(f"   Extra weekly revenue = ₦{improvement_gain:,.0f}")
print()

# What-if 3: All days matched the best day
best_week = highest * 5
print(f"3. If ALL 5 days matched best day (₦{highest:,.0f}):")
print(f"   Weekly revenue = ₦{best_week:,.0f}")
print(f"   vs actual ₦{total:,.0f}")
print(f"   Potential upside = ₦{best_week - total:,.0f}")
print("-" * 45)
```

```python
# ================================================
# SECTION 3: PERFORMANCE CHART WITH BENCHMARKS
# ================================================

plt.figure(figsize=(10, 5))

# Bar chart
plt.bar(df["date"], df["sales_amount"],
        color="steelblue", label="Daily Sales")

# Average line
plt.axhline(y=average, color="orange",
            linestyle="--", label=f"Average ₦{average:,.0f}")

# Median line
plt.axhline(y=median, color="green",
            linestyle="--", label=f"Median ₦{median:,.0f}")

# Top 20% line
plt.axhline(y=top_20_threshold, color="red",
            linestyle="--",
            label=f"Top 20% Threshold ₦{top_20_threshold:,.0f}")

plt.title("Daily Sales With Statistical Benchmarks — May 2026")
plt.xlabel("Date")
plt.ylabel("Sales in ₦")
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.savefig("sales_insight_chart.png")
plt.show()
print("Chart saved!")
```
