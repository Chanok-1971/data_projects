```python
import pandas as pd
import matplotlib.pyplot as plt

# Dataset Extraction
data = {
    'order_id': ['ORD-001','ORD-002','ORD-003','ORD-004','ORD-005',
                 'ORD-006','ORD-007','ORD-008','ORD-009','ORD-010',
                 'ORD-011','ORD-012','ORD-013','ORD-014','ORD-015'],
    'date': ['2026-05-01','2026-05-01','2026-05-02','2026-05-02','2026-05-03',
             '2026-05-03','2026-05-04','2026-05-04','2026-05-05','2026-05-05',
             '2026-05-06','2026-05-06','2026-05-07','2026-05-07','2026-05-08'],
    'amount': [12500,8750,22000,5500,18000,9200,31000,6800,14500,27500,
               8100,19800,42000,7300,15600],
    'product_category': ['Electronics','Clothing','Electronics','Food','Electronics',
                         'Clothing','Electronics','Food','Clothing','Electronics',
                         'Food','Clothing','Electronics','Food','Clothing'],
    'customer_country': ['Nigeria','Nigeria','Nigeria','Nigeria','Ghana',
                         'Ghana','Nigeria','Nigeria','Nigeria','Ghana',
                         'Nigeria','Nigeria','Nigeria','Ghana','Nigeria']
}

df = pd.DataFrame(data)

# Vectorized aggregation for Revenue Allocation Share per Country
# Mathematical mapping: \text{Share}_{country} = \frac{\sum_{j \in country} \text{amount}_j}{\sum \text{amount}}
country_rev = df.groupby('customer_country')['amount'].sum()

# Visual Compilation
fig, ax = plt.subplots()
country_rev.plot(
    kind='pie',
    ax=ax,
    autopct='%1.1f%%',
    startangle=140,
    colors=['steelblue', 'coral'],
    textprops={'weight': 'bold'}
)

ax.set_title('Revenue Allocation Share by Customer Country', fontsize=12, fontweight='bold')
ax.set_ylabel('')  # Suppress default vertical axis label alignment noise

plt.tight_layout()
plt.savefig('pie_chart.png')
```
